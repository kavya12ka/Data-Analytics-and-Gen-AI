### Historical Dataset key value

```python
# Last 252 trading days - Trading days in 1 year
last_52_weeks = historical_data.tail(252)

# Last 52 week high
week_52_high = (last_52_weeks["High"].max())

# Last 52 week low
week_52_low = (last_52_weeks["Low"].min())

one_year_return = ((historical_data['Close'].iloc[-1]/historical_data['Close'].iloc[-252]-1)*100)

print("52-Week High:",week_52_high)
print("52-Week Low:",week_52_low)
print("One Year Return:",one_year_return)
```

```python
three_year_return = (historical_data['Close'].iloc[-1]/historical_data['Close'].iloc[-756]-1)*100
print("Three Year Return:",three_year_return)
```

```python
five_year_date = historical_data.index.min()
five_year_date
```

```python
five_year_price = historical_data.loc[historical_data.index>=five_year_date,"Close"].iloc[0]
five_year_return = (historical_data['Close'].iloc[-1]/five_year_price-1)*100
print("Five Year Return:",five_year_return)
```

```python
historical_data["Daily Return"] = historical_data["Close"].pct_change()
annual_volatility = round(historical_data["Daily Return"].std()* (252 ** 0.5)* 100,2)
print("Annualized Volatility:", annual_volatility)
```

```python
historical_data["Peak"] = historical_data["Close"].cummax() #Cumulative Maximum
```

```python
historical_data["Drawdown"] = (historical_data["Close"]-historical_data["Peak"])/historical_data["Peak"]
historical_data[["Close","Peak","Drawdown"]]
```

```python
max_drawdown = (historical_data["Drawdown"].min())
max_drawdown_percentage = max_drawdown *100
max_drawdown_percentage
```

```python
initial_price = five_year_price
final_price = historical_data['Close'].iloc[-1] #last row
annualized_return = (final_price/initial_price)**(1/5)-1
annualized_return_percentage = annualized_return*100
print("Annualized Return Percentage:",annualized_return_percentage)
```

```python
sharpe_ratio = (annualized_return-0)/annual_volatility *100
print("Sharpe ratio:",sharpe_ratio)
```

### Embedding and storing vectors in FAISS


```python
EMBED_MODEL = "gemini-embedding-001"
GEN_MODEL   = "gemini-3.5-flash"

EMBED_DIM = 768

def embed(texts, task_type):
    out = []
    for i in range(0, len(texts), 50):
            res = client.models.embed_content(
                model=EMBED_MODEL, contents=texts[i:i+50],
                config=types.EmbedContentConfig(task_type=task_type, output_dimensionality=EMBED_DIM))
            out.extend([e.values for e in res.embeddings])
    return np.array(out,dtype = "float32")

chunk_texts = [f"{chunks['header']}. {chunks['text']}" for chunks in documents]
chunk_vecs  = embed(chunk_texts, "RETRIEVAL_DOCUMENT")
# Add index
index = faiss.IndexFlatIP(EMBED_DIM)
index.add(chunk_vecs)
print("Vectors in FAISS index:", index.ntotal)
```

# Retrieve relevent chunks and answers

```python
def retrieve(question, k=2):
    q = embed([question], "RETRIEVAL_QUERY")
    scores, ids = index.search(q, k)
    return [(documents[int(i)], float(s))
           for i, s in zip(ids[0], scores[0]) if 0<=i<len(documents)]

def ask(question, k=2, show_context=False):
    hits = retrieve(question, k)
    # answer for the question
    context = "\n\n".join(f"[{n}] {chunk['header']} (source: {chunk['source']})\n{chunk['text']}"
                          for n, (chunk, source) in enumerate(hits, 1))
    # prompt build
    prompt = f"CONTEXT:\n{context}\n\nQUESTION: {question}"
    #generate ans
    resp = client.models.generate_content(
        model=GEN_MODEL, contents=prompt)
    print("Question:", question)
    print("Retrieved:", [(chunk["header"], round(source, 3)) for chunk, source in hits])
    if show_context: print(context)
    print("\n" + resp.text + "\n" + "=" * 80)
    return resp.text
```

```python
test_questions = [
    "1.How has Microsoft's revenue and revenue growth changed over the years?",
    "2.Is Microsoft profitable? Discuss net income, margins and EPS.",
    "3.How strong is Microsoft's balance sheet? Discuss assets, liabilities, debt and cash.",
    "4.How much free cash flow does Microsoft generate and how much is returned to shareholders?"
]
for q in test_questions:
  ask(q)
```
